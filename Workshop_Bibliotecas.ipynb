```python
import numpy as np
temperaturas = np.round(np.random.uniform(20, 35, 7), 1)
print(temperaturas)
```

```python
import pandas as pd
import numpy as np

dias = np.array(['Segunda', 'Terça', 'Quarta', 'Quinta', 'Sexta', 'Sábado', 'Domingo'])

df = pd.DataFrame({
    'Dia': dias,
    'Temperatura': temperaturas
})

print(df)
```

```python
media_temperatura = df['Temperatura'].mean()
print("Média da temperatura da semana:", np.round(media_temperatura, 1))
```

```python
import matplotlib.pyplot as plt

plt.plot(df['Dia'], df['Temperatura'], marker = 'o', color = 'b')

plt.title('Variação da temperatura ao longo da Semana')
plt.xlabel('Dia da Semana')
plt.ylabel('Temperatura (°C)')

plt.show()
```
